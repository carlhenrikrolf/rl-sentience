```python
liking = [ # content-specific ncc for liking
    'ACB', # nucleus accumbens (should be the shell a separate hotspot and coldspot)
    'PALv', # ventral pallidum (most important)
    'PB', # parabrachial nucleus
]
wanting = [ # content-specific ncc for wanting
    'SNc', # substantia nigra pars compacta
    'VTA', # ventral tegmental area
]
# ncc = [ # full ncc
#     'TEa', # temporal (association areas)
#     'PTLp', # posterior parietal (association areas)
#     'VIS', # occiptal (visual areas)
# ]
```

```python
# Create a brainrender scene
from brainrender import Scene
scene = Scene(atlas_name='allen_mouse_50um', title='mouse')

# reinforcement (temporal difference)
settings = {'alpha': 0.67, 'silhouette': True, 'force': True}
liking_scene = scene.add_brain_region(*liking, color='green', **settings)
scene.add_label(liking_scene, 'liking')
wanting_scene = scene.add_brain_region(*wanting, color='cyan', **settings)
scene.add_label(wanting_scene, 'wanting')
# ncc_scene = scene.add_brain_region(*ncc, color='yellow', **settings)
# scene.add_label(ncc_scene, 'ncc')
```

```python
# Set the backend
import vedo
vedo.settings.default_backend= 'k3d'

# Make sure it gets embedded in the window
scene.jupyter = True

# scene.render now will prepare the scene for rendering, but it won't render anything yet
scene.render(camera='frontal') # frontal, sagittal

# To display the scene we use `vedo`'s `show` method to show the scene's actors
# from vedo import Plotter  # <- this will be used to render an embedded scene 
plt = vedo.Plotter()
plt.show(*scene.renderables)  # same as vedo.show(*scene.renderables)                                                                                   
```
